```python
import pandas as pd

df1 = pd.read_csv("store_sales_1.csv")
df2 = pd.read_csv("store_sales_2.csv")
df3 = pd.read_csv("store_sales_3.csv")

# Combine all 3 into one DataFrame
df_all = pd.concat([df1, df2, df3], ignore_index=True)

df_all
```

```python
exchange_rates = {
    "USD": 0.385,   # 1 USD = 0.385 OMR
    "SAR": 0.1027,  # 1 SAR = 0.1027 OMR
    "OMR": 1.0      # OMR يبقى نفسه
}
df_all["Unit_Price"] = df_all.apply(
    lambda row: row["Unit_Price"] * exchange_rates.get(row["CurrencyType"], 1.0),
    axis=1
)
df_all["CurrencyType"] = "OMR"
df_all
```

```python
# ----------------------------------
# Create Customers table if it doesn't exist
# ----------------------------------
create_customers_table_query = """
CREATE TABLE IF NOT EXISTS customers (
    customer_id VARCHAR(36) PRIMARY KEY,
    UNIQUE KEY unique_customer (customer_id)
);
"""

cursor.execute(create_customers_table_query)
connection.commit()
print("Table 'customers' is ready!")
```

```python
# ----------------------------------
# Create Products table if it doesn't exist
# ----------------------------------
create_products_table_query = """
CREATE TABLE IF NOT EXISTS products (
    product_id INT AUTO_INCREMENT PRIMARY KEY,
    product_name VARCHAR(255) ,
    unit_price DECIMAL(10,4) ,
    qty INT,
    
    UNIQUE KEY unique_product (product_name)
);
"""

cursor.execute(create_products_table_query)
connection.commit()
print("Table 'products' is ready!")
```

```python
# ----------------------------------
# Create Stores table if it doesn't exist
# ----------------------------------
create_stores_table_query = """
CREATE TABLE IF NOT EXISTS stores (
    store_id VARCHAR(50) PRIMARY KEY,
    UNIQUE KEY unique_store (store_id)
);
"""

cursor.execute(create_stores_table_query)
connection.commit()
print("Table 'stores' is ready!")
```

```python
create_sales_table_query = """
CREATE TABLE IF NOT EXISTS sales_info (
    sale_id INT AUTO_INCREMENT PRIMARY KEY,
    product_id INT,
    qty INT,
    unit_price DECIMAL(10,4),
    sale_date DATE,
    currency_type VARCHAR(10),
    customer_id CHAR(36),
    store_id VARCHAR(50),
    total_price DECIMAL(10,2),

    UNIQUE KEY unique_sale (product_id, customer_id, store_id, sale_date),

    FOREIGN KEY (customer_id) REFERENCES customerS(customer_id),
    FOREIGN KEY (store_id) REFERENCES stores(store_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
"""
cursor.execute(create_sales_table_query)
connection.commit()
print("Table 'sales_info' created successfully!")
```

```python
from tqdm import tqdm

insert_query = """
INSERT IGNORE INTO products (product_name, unit_price, qty)
VALUES (%s, %s, %s);
"""

for _, row in tqdm(df.iterrows(), total=df.shape[0], desc="Inserting products"):
    product_name = row["ProductName"]
    unit_price = row["Unit_Price"] if pd.notna(row["Unit_Price"]) else None
    qty = row["Qty"] if pd.notna(row["Qty"]) else None

    cursor.execute(insert_query, (product_name, unit_price, qty))

connection.commit()
print("Products inserted successfully!")

```

```python
from tqdm import tqdm
import pandas as pd

# Assuming df is your DataFrame with a 'StoreID' column

insert_query = """
INSERT IGNORE INTO stores (store_id)
VALUES (%s);
"""

for _, row in tqdm(df.iterrows(), total=df.shape[0], desc="Inserting stores"):
    store_id = row["StoreID"]
    
    # Convert NaN to None so MySQL can accept it
    if pd.isna(store_id):
        store_id = None

    cursor.execute(insert_query, (store_id,))

connection.commit()
print("Stores inserted successfully!")
```

```python
for _, row in tqdm(df_all.iterrows(), total=df_all.shape[0], desc="Inserting sales"):
    # Get ProductID
    cursor.execute("SELECT product_id FROM products WHERE product_name=%s", (row['ProductName'],))
    result = cursor.fetchone()
    if result is None:
        continue
    pid = result[0]

    cursor.execute("""
        INSERT IGNORE INTO sales_info
        (product_id, qty, unit_price, sale_date, currency_type, customer_id, store_id, total_price)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
    """, (
        pid,
        row['Qty'],
        row['Unit_Price'],
        row['SaleDate'],
        row['CurrencyType'],
        row['CustomerID'],
        row['StoreID'],
        row['TotalPrice']
    ))

connection.commit()
connection.close()
print("Data successfully inserted into MySQL!")
```
