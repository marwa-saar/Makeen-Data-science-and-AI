```python
import numpy as np
```

```python
scores=[50,60,60,70,70,70,80,80,90]
```

```python
mean=np.mean(scores)
mean
```

```python
std=np.std(scores)
std
```

```python
z_scores=[(x-mean)/std for x in scores]
z_scores
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.hist(z_scores, bins=5, color='green', edgecolor='red')
plt.title("")
plt.xlabel("")
plt.ylabel("")
plt.locator_params(axis='y',integer=True)# number stu int
plt.show()
```
