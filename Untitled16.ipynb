```python
import pymysql
from prefect import flow, task
import pandas as pd
```

```python



# --------------------- TASKS ---------------------

# EXTRACT
@task
def extract():
    df = pd.read_csv(
        "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
    )
    return df


@task
def transform(df: pd.DataFrame) -> pd.DataFrame:
    df['Age'] = df['Age'].fillna(df['Age'].mean()).astype(int)
    df['Fare'] = df['Fare'].fillna(df['Fare'].mean()).astype(float)
    df = pd.get_dummies(df, columns=['Sex'], drop_first=True)
    print("Data transformed")
    return df

@task
def load(df: pd.DataFrame, host, user, password, database, table_name: str):
    conn = pymysql.connect(host=host, user=user, password=password, database=database)
    cursor = conn.cursor()
    
    # Create table if not exists
    cursor.execute(f"""
        CREATE TABLE IF NOT EXISTS pass (
            PassengerId INT PRIMARY KEY,
            Survived INT,
            Pclass INT,
            Name VARCHAR(100),
            Age INT,
            Fare FLOAT,
            Sex_male TINYINT
        )
    """)
    
    # Insert data row by row
    for _, row in df.iterrows():
        placeholders = ', '.join(['%s'] * len(row))
        columns = ', '.join(row.index)
        sql = f"REPLACE INTO pass ({columns}) VALUES ({placeholders})"
        cursor.execute(sql, tuple(row))
    
    conn.commit()
    cursor.close()
    conn.close()
    print(f"Loaded {len(df)} rows into pass")

# --------------------- FLOW ---------------------

@flow
def etl_flow(host, user, password, database, table_name):
    csv_file =train()
    df = extract(train)
    df_transformed = transform(df)
    load(df_transformed, host, user, password, database, table_name)

# --------------------- RUN ---------------------

if __name__ == "__main__":
    etl_flow(
        host="localhost",
        user="root",
        password="Marwa@3425",
        database="taitan",
        table_name="pass"
    )
```
