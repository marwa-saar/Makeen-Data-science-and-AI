```python
import pandas as pd
import glob
import pymysql
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
```

```python

# -----------------------------
# 1. EXTRACT
# -----------------------------

df1 = pd.read_csv("store_sales_1.csv")
df2 = pd.read_csv("store_sales_2.csv")
df3 = pd.read_csv("store_sales_3.csv")

# Combine all 3 into one DataFrame
df= pd.concat([df1, df2, df3], ignore_index=True)

df
# -----------------------------
```

```python
# 2. TRANSFORM
# -----------------------------
def fill_numeric(df):
    df = df.copy()
    df["Qty"] = df["Qty"].fillna(0).astype(int)
    df["Unit_Price"] = df["Unit_Price"].fillna(0.0).astype(float)
    return df

def fill_customers(df):
    df = df.copy()
    df["CustomerID"] = df["CustomerID"].fillna("GUEST")
    return df

def convert_date(df):
    df = df.copy()
    df["SaleDate"] = pd.to_datetime(df["SaleDate"])
    return df

def calc_total(df):
    df = df.copy()
    df["Total_Price"] = df["Qty"] * df["Unit_Price"]
    return df

def convert_currency(df):
    df = df.copy()
    rates = {"OMR": 1, "USD": 0.385, "EUR": 0.42}
    df["Total_OMR"] = df.apply(lambda r: r["Total_Price"] * rates.get(str(r["CurrencyType"]).upper(), 1), axis=1)
    return df
```

```python




pipeline = Pipeline([
    ("fill_numeric", FunctionTransformer(fill_numeric)),
    ("fill_customers", FunctionTransformer(fill_customers)),
    ("convert_date", FunctionTransformer(convert_date)),
    ("calc_total", FunctionTransformer(calc_total)),
    ("convert_currency", FunctionTransformer(convert_currency)),
])

df_transformed = pipeline.fit_transform(df)
print("Transformation Complete!")
print(df_transformed.head())

# -----------------------------
```

```python
from tqdm import tqdm
import pandas as pd

# Convert SaleDate properly
df_transformed['SaleDate'] = pd.to_datetime(df_transformed['SaleDate'], errors='coerce')

# Replace all NaN with None for SQL compatibility
df_transformed = df_transformed.where(pd.notnull(df_transformed), None)

for _, row in tqdm(df_transformed.iterrows(), total=df_transformed.shape[0], desc="Inserting Sales"):
    
    # ----- Get ProductID -----
    cursor.execute("SELECT ProductID FROM Product WHERE ProductName=%s", (row['ProductName'],))
    result = cursor.fetchone()

    if result is None:
        print("❌ Product not found:", row['ProductName'])
        continue

    product_id = result[0]

    # ----- INSERT -----
    cursor.execute("""
        INSERT INTO Sales
        (SaleDate, StoreID, ProductID, CustomerID, Qty, Unit_Price, CurrencyType, Total_Price, Total_OMR)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
    """, (
        row['SaleDate'].date() if row['SaleDate'] else None,
        row['StoreID'],
        product_id,
        row['CustomerID'],
        row['Qty'],
        row['Unit_Price'],
        row['CurrencyType'],
        row['Total_Price'],
        row['Total_OMR']
    ))

connection.commit()
connection.close()
print("✅ Sales inserted successfully!")
```
