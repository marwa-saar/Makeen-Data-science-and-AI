```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer

def fill_missing(df):
    df=df.copy()
    df['Age']=df['Age'].fillna(df['Age'].median())
    df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
    return df

def encoding(df):
    df=df.copy()
    gender={'male':1,'female':0}
    Embarked_map={'C':0,'S':1,'Q':2}
    df['Sex']=df['Sex'].map(gender)
    df['Embarked']=df['Embarked'].map(Embarked_map)
    return df
    
def scale_numeric(df):
    df=df.copy()
  ### from sklearn.preprocessing import StandardScaler
    scaler=StandardScaler()
    df[['Age','Fare']]=scaler.fit_transform(df[['Age','Fare']])
    return df

pipeline=Pipeline([('fill_missing',FunctionTransformer(fill_missing)),
                   ('encoding',FunctionTransformer(encoding)),
                   ('scale_numeric',FunctionTransformer(scale_numeric)),])
df_transformed=pipeline.fit_transform(df)
df_transformed
```

```python
# --- CLEAN SEX COLUMN ---
df_transformed["Sex"] = df_transformed["Sex"].fillna("female")
df_transformed["Sex"] = df_transformed["Sex"].map({"male": 1, "female": 0})

# --- CLEAN EMBARKED COLUMN ---
df_transformed["Embarked"] = df_transformed["Embarked"].fillna("S")
df_transformed["Embarked"] = df_transformed["Embarked"].map({"S": 0, "C": 1, "Q": 2})

# --- CLEAN NUMERIC COLUMNS ---
df_transformed["Age"] = df_transformed["Age"].fillna(df_transformed["Age"].median())
df_transformed["Fare"] = df_transformed["Fare"].fillna(df_transformed["Fare"].median())

# --- ENSURE CORRECT DATA TYPES ---
df_transformed = df_transformed.astype({
    "PassengerId": int,
    "Survived": int,
    "Pclass": int,
    "Sex": int,
    "Embarked": int
})
```
