```python
import mysql.connector

def collect_data():
    # Connect to MySQL database (change these parameters accordingly)
    conn = mysql.connector.connect(
        host='localhost',          # or your MySQL server address
        user='root',      # your MySQL username
        password='Marwa@3425',  # your MySQL password
        database='MovieDB'         # your database name (as in your script)
    )

    cursor = conn.cursor()

    # Example query to collect all data from ProductionCompany table
    query = "SELECT * FROM ProductionCompany"

    cursor.execute(query)

    # Fetch all rows
    rows = cursor.fetchall()

    # Print rows or process them
    for row in rows:
        print(row)

    # Close cursor and connection
    cursor.close()
    conn.close()

if __name__ == "__main__":
    collect_data()
```
