```python
import requests
import json
import pandas as pd
```

```python
api_url = "https://fakerapi.it/api/v2/addresses?_quantity=30"
try:
    # 2. Make the GET request to the API
    response = requests.get(api_url)
    
    # 3. Raise an exception for bad status codes (4xx or 5xx)
    response.raise_for_status()

    # 4. Parse the JSON response
    data = response.json()

    print(f"✅ Successfully retrieved {data['total']} addresses.\n")
    
    # 5. Print the first 5 addresses as an example
    print("--- First 5 Retrieved Addresses ---")
    
    # The 'data' field in the response contains the list of addresses
    for i, address in enumerate(data.get('data', [])):
        if i < 5:
            # Use json.dumps for a nicely formatted printout
            print(f"\nAddress {i + 1}:")
            print(json.dumps(address, indent=4))
        else:
            break
            
except requests.exceptions.RequestException as e:
    # Handle any request errors (network, timeout, HTTP errors)
    print(f"❌ An error occurred while fetching data: {e}")
```

```python
import requests
import pandas as pd

api_url = "https://fakerapi.it/api/v2/addresses?_quantity=30"

try:
    # Make GET request
    response = requests.get(api_url, timeout=10)
    response.raise_for_status()

    # Parse JSON
    data = response.json()
    addresses = data.get('data', [])

    # Convert list of dictionaries to a DataFrame
    df = pd.DataFrame(addresses)

    print(f"✅ Successfully retrieved {len(df)} addresses.\n")

    # Show first 5 rows
    print("--- First 5 Addresses ---")
    print(df.head())

except requests.exceptions.RequestException as e:
    print(f"❌ An error occurred while fetching data: {e}")
```

```python
import requests
import pandas as pd

# 1. Fetch data from the API
api_url = "https://fakerapi.it/api/v2/addresses?_quantity=30"
response = requests.get(api_url)
response.raise_for_status()  # Raise error if request failed
data = response.json()

# 2. Convert the 'data' list to a DataFrame
addresses = data.get('data', [])
df = pd.DataFrame(addresses)

# 3. Save DataFrame to Excel
df.to_excel("addresses.xlsx", index=False)  # index=False avoids adding row numbers

print("✅ Data saved to 'addresses.xlsx'")
```
