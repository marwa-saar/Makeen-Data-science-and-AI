```python
# ----------------------------------
# Load Excel file
# ----------------------------------
excel_file = r"C:\Users\USER PC\OneDrive\Desktop\advance python\addresses.xlsx"
df = pd.read_excel(excel_file)

print("Excel file loaded successfully!")
df.head()
```

```python
# ----------------------------------
# Connect to MySQL
# ----------------------------------
connection = pymysql.connect(
    host="localhost",
    user="root",
    password="Marwa@3425",   # <-- change this
    database="oman_db"          # <-- your database name
)

cursor = connection.cursor()
print("Connected to MySQL!")
```

```python
# ----------------------------------
# Create table if it doesn't exist
# ----------------------------------
create_table_query = """
CREATE TABLE IF NOT EXISTS oman (
    id INT AUTO_INCREMENT PRIMARY KEY,
    street VARCHAR(255),
    streetName VARCHAR(255),
    buildingNumber VARCHAR(50),
    city VARCHAR(255),
    zipcode VARCHAR(50),
    country VARCHAR(255),
    country_code VARCHAR(10),
    latitude DOUBLE,
    longitude DOUBLE,
    UNIQUE KEY unique_address (street, buildingNumber, city)
);
"""

cursor.execute(create_table_query)
connection.commit()
print("Table 'oman' is ready!")
```

```python
# ----------------------------------
# Insert Excel data into MySQL with duplicate check
# ----------------------------------
insert_query = """
INSERT IGNORE INTO oman 
(street, streetName, buildingNumber, city, zipcode, country, country_code, latitude, longitude)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s);
"""

# Use tqdm for progress bar
for _, row in tqdm(df.iterrows(), total=df.shape[0], desc="Inserting rows"):
    cursor.execute(insert_query, (
        row.get("street"),
        row.get("streetName"),
        row.get("buildingNumber"),
        row.get("city"),
        row.get("zipcode"),
        row.get("country"),
        row.get("country_code"),
        row.get("latitude"),
        row.get("longitude")
    ))

connection.commit()
connection.close()
print("Data successfully inserted into 'oman' table!")
```
